```python
import os
import pandas as pd
import json
import subprocess
from datetime import datetime
import re
from hashlib import md5
```

```python
JSON_PATH = r"\\192.168.20.101\webdata\Nav\old"
OUTPUT_PATH = "./output/processd_json"
if not os.path.exists(OUTPUT_PATH):
    os.makedirs(OUTPUT_PATH)
```

```python
json_files = [
    os.path.join(JSON_PATH, file)
    for file in os.listdir(JSON_PATH)
    if file.endswith(".json")
]
```

```python
len(json_files)
```

```python
_jf = json_files[0]
```

```python
file_name = os.path.basename(_jf)
file_name
```

```python

centre_lat = -19.041845946974913
centre_lon = 50.39130509789334
f_count = 0
for _jf in json_files:
    with open(_jf, "r") as f:
        data = json.load(f)
        _df = pd.DataFrame(data)
        try:
            _df["latitude"] = _df["latitude"].astype(float) + centre_lat
        except:
            _df["latitude"] = "Not found"
        try:
            _df["longitude"] = _df["longitude"].astype(float) + centre_lon
        except:
            _df["longitude"] = "Not found"
        _df["equipment"] = _df["googleAccount"].apply(lambda x :md5(x.encode('utf-8')).hexdigest())
        _df = _df[
            [
                "latitude",
                "longitude",
                "equipment",
                "batteryLevel",
                "lastUpdate",
                "serverTime",
            ]
        ]
        _df["lat,long"] = _df.apply(
            lambda row: f"{row['latitude']},{row['longitude']}",
            axis=1
        )
        with open(os.path.join(OUTPUT_PATH, os.path.basename(_jf)), "w", encoding="utf-8") as f_out:
            f_out.write(
                _df.to_json(orient="records", force_ascii=False)
                .replace("\\/", "/"))
    f_count += 1   
```
